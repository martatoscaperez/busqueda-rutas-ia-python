#### Estrategias de búsqueda a implementar: Coste Uniforme, Primero el Mejor y A$^*$

La principal diferencia entre estos tres algoritmos radica en la forma en que se calcula la función de coste para un nodo específico $n$, es decir, $f(n) = g(n) + h(n)$. Suponiendo que $g(n)$ es el coste real acumulado desde el **estado inicial** hasta `n.state` y que $h(n)$ es el coste heurístico desde el estado `n.state` hasta el **estado objetivo**, $f(n)$ se calcula de la siguiente manera:

- Coste uniforme: $f(n) = g(n)$
- Mejor primero: $f(n) = h(n)$
- A$^*$: $f(n) = g(n) + h(n)$

##### Notas:

* El orden de las acciones se determina por el estado de destino cuyo identificador sea menor. Es decir, si se pueden alcanzar diferentes destinos (parciales) en un punto dado (ciudad), se visitarán en orden alfanumérico creciente.
* Para implementar la lista de **nodos abiertos**, se recomienda encarecidamente utilizar una estructura de datos de [Cola de Prioridad](https://docs.python.org/3/library/queue.html#queue.PriorityQueue), ya que es una estructura eficiente para ordenar los nodos, en lugar de tener que ordenar la lista cada vez. De igual forma, para implementar la lista de **nodos cerrados**, se recomienda encarecidamente utilizar una estructura de datos tipo *hash* por cuestiones de eficiencia.

## 6. Funciones para la creación y visualización de mapas

La siguiente función genera un mapa (grafo no dirigido) que representa un mapa de ciudades y sus conexiones (carreteras). Las ciudades reciben el nombre C1, C2, .... y todas las carreteras o enlaces son de doble sentido.

Para crear el mapa debemos pasarle los siguientes parámetros:

- `seed`: es una semilla para que la generación del grafo sea reproducible.
- `n_cities`: el número de ciudades que tendrá nuestro grafo/mapa.
- `max_neighbours`: el número máximo de ciudades a las que una ciudad puede estar conectada con un enlace (carretera) directo.
- `radious`: un valor $r \in (0,1]$. Indicará que $r\cdot 500$ es la distancia máxima entre dos ciudades para que potencialmente puedan estar conectadas. Es decir, si la distancia es mayor nunca estarán conectadas, pero si es menor, no se garantiza la conexión (dependerá de `max_neigbours`).
  
El resultado es un mapa en una cuadrícula $[0,500]\times[0,500]$ con las ciudades en posiciones (x,y) generadas aleatoriamente y sus conexiones directas. Las estructuras de datos que almacenan esta información son:
* cities: List[Dict]. Lista de ciudades. Cada nodo es un diccionario con claves: 'name', 'x', 'y'.
* roads : Dict[str, List[Dict]]. Diccionario de adyacencias (carreteras o conexiones) indexado por nombre de ciudad. Cada valor es una lista de enlaces, donde cada enlace es un dict con claves: 'to' (ciudad del nodo destino) y 'distance' (float).


```python
# Importamos librerías
import random
import math
from pprint import pprint
from typing import List, Dict, Tuple

import matplotlib.pyplot as plt
import networkx as nx


# Función para generar un grafo según las restricciones indicadas 

def generate_map(seed: int,
                  n_cities: int = 20,
                  max_neighbours: int = 5,
                  radious: float = 10.0
                 ) -> Tuple[List[Dict], Dict[str, List[Dict]]]:
    """
    Genera un mapa de ciudades y carreteras (grafo no dirigido) sobre un plano 2D [0,500]x[0,500].

    Comportamiento
    --------------
    - Las ciudades se nombran como C1, C2, ..., Cn.
    - Las coordenadas (x, y) se generan ~ U[0, 500] (valores enteros).
    - Se crea un enlace no dirigido entre dos ciudades si la distancia < radiousE 
    - Si un nodo supera 'max_neighbours', se reducen enlaces aleatoriamente manteniendo la no direccionalidad.
    """

    # =============================
    # Configuración y validaciones
    # =============================
    if n_cities <= 0:
        raise ValueError("n_cities debe ser un entero positivo.")
    if max_neighbours <= 0:
        raise ValueError("max_neighbours debe ser un entero > 0.")
    if radious < 0 or radious >=1:
        raise ValueError("radious debe ser > 0.")
    else:
        radiousA = 500*radious

    random.seed(seed)

    
    # =============================
    # Generación de ciudades
    # =============================
    cities: List[Dict] = []
    for i in range(1, n_cities + 1):
        cities.append({
            "name": f"C{i}",
            "x": random.randint(0, 500),
            "y": random.randint(0, 500)
        })

    # Índices auxiliares: nombre -> índice y viceversa
    name_to_idx = {n["name"]: idx for idx, n in enumerate(cities)}
    idx_to_name = {idx: n["name"] for idx, n in enumerate(cities)}

    # =============================
    # Cálculo de enlaces por radious
    # =============================
    # Usamos un set de pares (i, j) i<j para evitar duplicados.
    roads = set()
    distances = {}  # clave (i, j) con i<j -> distancia

    for i in range(n_cities):
        xi, yi = cities[i]["x"], cities[i]["y"]
        for j in range(i + 1, n_cities):
            xj, yj = cities[j]["x"], cities[j]["y"]
            d = math.sqrt((xi - xj)*(xi-xj) + (yi - yj)*(yi-yj))
            d = d*(1+random.uniform(0.0, 0.1)) #se suma ruido para simular que la carretera no es una recta
            if d < radiousA:
                roads.add((i, j))
                distances[(i, j)] = d

    # Construimos la carretera (no dirigida) inicial
    road: Dict[str, List[Dict]] = {n["name"]: [] for n in cities}
    for (i, j) in roads:
        ni, nj = idx_to_name[i], idx_to_name[j]
        d = distances[(i, j)]
        road[ni].append({"to": nj, "distance": d})
        road[nj].append({"to": ni, "distance": d})

    # =============================
    # Eliminamos carreteras para asegurar el número máximo de vecinos
    # =============================
    # Para evitar sesgos, iteramos las ciudades en orden aleatorio.
    city_order = list(idx_to_name.values())
    random.shuffle(city_order)
    
    for name in city_order:
        neighbours = road[name]
        if len(neighbours) > max_neighbours:
            # Seleccionamos aleatoriamente qué vecinos mantener
            to_keep = set(v["to"] for v in random.sample(neighbours, max_neighbours))
            to_remove = [v for v in neighbours if v["to"] not in to_keep]

            # Eliminamos en ambos sentidos para mantener el grafo no dirigido
            for v in to_remove:
                target = v["to"]
                # Eliminar enlace name -> target
                road[name] = [w for w in road[name] if w["to"] != target]
                # Eliminar enlace target -> name
                road[target] = [w for w in road[target] if w["to"] != name]

    return cities, road
```

The following method draws the generated map. In addition:
* If `path` contains a path between cities, it highlights the links between them (if any).
* If `path` is `None`, it doesn't highlight any links.

**Important**: The method doesn't check that the path actually exists; it must be verified before calling it.

El siguiente método dibuja el mapa generado. Además:
* Si `path`contiene un camino entre ciudades, resalta los enlaces entre ellas (si existen).
* Si `path`es `None` no resalta ningun enlace.

**Importante**: El método no chequea que el camino realmente exista, debe verificarse antes de lanzar la llamada. 

```python
# ======================================
# Dibujar el mapa y opcionalmente resaltar un camino 
# ======================================

def draw_map(cities: List[Dict],
             road: Dict[str, List[Dict]],
             path: List[str],
             figsize=(8, 8),
             show_distances=False):
    """
    Dibuja el grafo y resalta el camino dado (lista de nombres de nodos).
    """
    # Construimos el grafo y posiciones
    G = nx.Graph()
    pos = {}
    for n in cities:
        name = n["name"]
        pos[name] = (float(n["x"]), float(n["y"]))
        G.add_node(name)

    added = set()
    for u, neighbours in road.items():
        for v in neighbours:
            w = v["to"]
            d = float(v["distance"])
            key = tuple(sorted((u, w)))
            if key not in added:
                G.add_edge(u, w, weight=d)
                added.add(key)

    # Dibujo base
    plt.figure(figsize=figsize)
    nx.draw(G, pos, node_size=500, node_color="#89CFF0",
            edge_color="#bbbbbb", with_labels=True, font_size=9, width=1.4)

    # Resaltar camino
    if path and len(path) > 1:
        path_edges = list(zip(path[:-1], path[1:]))
        nx.draw_networkx_nodes(G, pos, nodelist=path, node_color="#FFD166", node_size=600)
        nx.draw_networkx_edges(G, pos, edgelist=path_edges, width=3.0, edge_color="#EF476F")
        # Opcional: etiquetas de distancia solo en las aristas del camino
        if show_distances:
            edge_labels = {(u, v): f'{G[u][v]["weight"]:.2f}' for u, v in path_edges}
            nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8)

    plt.axis("equal"); plt.xlim(0, 520); plt.ylim(0, 520)
    plt.tight_layout()
    plt.show()
```

```python
from queue import PriorityQueue
from typing import Dict
import math

class Estado:
    def __init__(self, identificador: str, longitud: float, latitud: float): 
        self.identificador= identificador 
        self.longitud= longitud
        self.latitud = latitud

    def __eq__ (self, other): 
        if not isinstance(other,Estado):
            return False
        return self.identificador==other.identificador
    
    def __hash__(self): 
        return hash(self.identificador) 
                
    def __str__(self): 
        return f"Ciudad: {self.identificador}, Coordenadas: {self.longitud},{self.latitud}"
    
    def __repr__(self): 
        return f"Estado({self.identificador}, {self.longitud}, {self.latitud})"


class Accion:
    def __init__(self, origen: str, destino: str, distancia: float):
        self.origen = origen 
        self.destino = destino
        self.coste = distancia 
        
    def __str__(self):
        return f"{self.origen}-->{self.destino}, Distancia: {self.coste}"
    
    def __repr__(self):
        return f"Accion({self.origen}, {self.destino}, {self.coste})"
    
class Nodo: 
    def __init__(self, estado, padre=None, accion=None):
               
        self.estado=estado 
        self.padre=padre
        self.accion=accion 
        if padre is None: 
            self.coste=0
            self.profundidad=0
        else: 
            self.coste=padre.coste + (accion.coste if accion else 0)
            self.profundidad= padre.profundidad + 1
            
            
    
    def __lt__(self,other): #en caso de empate se usara un contador para saber que nodo se genero primero (meteremos el primero en la lista de abiertos)
        return self.coste < other.coste
    
    def __repr__(self):
        return f"Nodo({self.estado},{self.padre},{self.accion},{self.coste},{self.profundidad})"
    
    def __str__(self):
        return f"Nodo: {self.estado.identificador}, padre: {self.padre}, accion: {self.accion}, coste: {self.coste:.2f}, profundidad: {self.profundidad})"
   

class Problema:
    def __init__(self, cities, roads, origen_id, destino_id):      
        self.estados: Dict[str, Estado]= { c["name"]: Estado(c["name"], c["x"], c["y"]) for c in cities}
        self.roads=roads   
        self.inicial: Estado= self.estados[origen_id]            
        self.objetivo: Estado= self.estados[destino_id]             
        
    def estado_objetivo(self, estado: Estado):
        return estado.identificador==self.objetivo.identificador   


class BusquedaAlgoritmo:
    def __init__(self, problema: Problema):
        self.problema=problema 
        self.cerrados=set() #conjunto de estados ya expandidos
        self.n_generados=0
        self.n_expandidos=0
        self.n_expandidos=0
        self.profundidad_solucion=None 
        
    def frontera(self): 
        self.contador=0
        return PriorityQueue()
    
    def insertar_nodo(self, nodo): 
        return None
        
    def extraer_nodo(self):
        return self.abiertos.get()[2]  
        
    def esta_vacio(self): 
        return self.abiertos.empty()
       

    def expandir(self,nodo:Nodo):
        sucesores=[]
        for t in sorted(self.problema.roads.get(nodo.estado.identificador,[]), key=lambda e:e["to"]):#comprueba si id del estado no esta presente como clave en diccionario roads
            destino=t["to"]
            coste=t["distance"]
            #parametros del hijo
            estado_hijo=self.problema.estados[destino]
            accion= Accion(nodo.estado.identificador, destino,coste)
            ##creo nodo hijo
            n_hijo=Nodo(estado_hijo,nodo,accion)
            sucesores.append(n_hijo)
            
        return sucesores

    def buscar(self):
        #--REINICIO POR EJECUCION-- 
        self.abiertos=self.frontera() 
        self.cerrados.clear()
        self.n_expandidos=0
        self.n_generados=0
        self.n_explorados=0
        self.profundidad_solucion=None
      
        self.insertar_nodo(Nodo(self.problema.inicial))
        self.n_generados +=1
        
        while not self.esta_vacio():
            actual=self.extraer_nodo() 
            if actual.estado in self.cerrados:
                self.n_explorados +=1
                continue            
            self.cerrados.add(actual.estado)
            self.n_explorados +=1 
            if self.problema.estado_objetivo(actual.estado):             
                self.profundidad_solucion=actual.profundidad
                return self.reconstruir_camino(actual)        
            
            sucesores= self.expandir(actual)
            self.n_expandidos +=1 
            for hijo in sucesores:
                    self.insertar_nodo(hijo) 
                    self.n_generados +=1
        #No hay solucion
        return None
    
    def heuristica(self,estado):
        x=estado.longitud - self.problema.objetivo.longitud 
        y=estado.latitud  - self.problema.objetivo.latitud 
        return math.hypot(x,y) 
    
    def reconstruir_camino(self,nodo: Nodo):
        camino=[]
        while nodo and nodo.accion is not None: 
            camino.append(nodo.accion) 
            nodo=nodo.padre 
        camino.reverse()
        return camino
    
    
    ###ESTRATEGIA COSTE UNIFORME###
class CosteUniforme(BusquedaAlgoritmo): #f(n)=g(n)
    def insertar_nodo(self, nodo):
        self.abiertos.put((nodo.coste, self.contador, nodo))
        self.contador +=1
        
    ###ESTRATEGIA PRIMERO EL MEJOR###
class PrimeroMejor(BusquedaAlgoritmo): #f(n)=h(n)
    def insertar_nodo(self, nodo):
        f = self.heuristica(nodo.estado)
        self.abiertos.put((f, self.contador, nodo))
        self.contador += 1

    ###ESTRATEGIA ALGORITMO-ESTRELLA###
class AEstrella(BusquedaAlgoritmo):  #f(n)=g(n)+h(n)
    def insertar_nodo(self, nodo):
        f = nodo.coste + self.heuristica(nodo.estado)
        self.abiertos.put((f, self.contador, nodo))
        self.contador += 1        
```

```python
import time
# Convierte List[Accion] -> List[str] (ids de ciudades), que es lo que espera draw_map
def acciones_a_nodos(camino):
    return [camino[0].origen] + [a.destino for a in camino] if camino else []

def imprimir_camino(acciones):
    if not acciones:
        print("Sin solución")
        return
    total = 0.0
    for a in acciones:
        print(f"  {a.origen} --> {a.destino}")
        total += a.coste
    print(f"  Coste total: {total:.2f}")
    
def run_algo(AlgClase, problema, nombre, dibujar=True,repeticiones=10000):
    buscador = AlgClase(problema)
    acciones = buscador.buscar() or []
    
    #para mostrar el tiempo medio de ejecucion de cpu
    t0 = time.perf_counter()
    for _ in range(repeticiones):
        buscador_tmp = AlgClase(problema)
        buscador_tmp.buscar()
    t1 = time.perf_counter()

    tiempo_medio = (t1 - t0) / repeticiones

    # Métricas en el main 
    coste_total = sum(a.coste for a in acciones) if acciones else float("inf")
    profundidad = buscador.profundidad_solucion
    explorados = getattr(buscador, "n_explorados", len(buscador.cerrados))

    metricas = {
        "generados": buscador.n_generados,
        "expandidos": buscador.n_expandidos,
        "explorados": explorados,                 
        "profundidad": profundidad,
        "coste": round(coste_total,2), #para que imprima solo dos decimales
        "tiempo_cpu_segundos": round(tiempo_medio, 6) #tiempo en segundos con 6 decimales
    }

    # Path de ids para draw_map
    path_ids = acciones_a_nodos(acciones)

    # Salida por consola
    print(f"\n[{nombre}]")
    print("Métricas:", metricas)
    imprimir_camino(acciones)

    # Dibujo 
    if dibujar:
        try:
            
            draw_map(cities_res, road_res, path_ids, show_distances=True)  
        except NameError:
            
            pass

    return path_ids, acciones, metricas


if __name__ == "__main__":
    cities_res, road_res = generate_map(seed=142, n_cities=50, max_neighbours=10, radious=0.2)
    problema = Problema(cities_res, road_res, "C28", "C5")
    
 # Ejecutar los algoritmos
    run_algo(CosteUniforme, problema, "Coste Uniforme")
    run_algo(PrimeroMejor, problema, "Primero el Mejor (Greedy)")
    run_algo(AEstrella, problema, "A*")
    
    
    
```

```python
import statistics
from time import perf_counter as time
import random




def escenario_1_mapas_moderados():
    """
    Escenario 1:
    - 10 mapas de tamaño moderado
    - 40 ciudades
    - conectividad media
    - Origen = C1, Destino = C40
    """

    N_MAPAS = 10
    seeds = [1000 + i for i in range(N_MAPAS)]

    coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu = ([] for _ in range(6))
    coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp = ([] for _ in range(6))
    coste_a,  prof_a,  gen_a,  exp_a,  expl_a,  t_a  = ([] for _ in range(6))

    for seed in seeds:
        random.seed(seed)

        cities, roads = generate_map(
            seed=seed,
            n_cities=40,
            max_neighbours=6,   # conectividad media
            radious=0.15        # radio medio
        )

        problema = Problema(cities, roads, "C1", "C40")

        # Coste Uniforme
        buscador = CosteUniforme(problema)
        t0 = time(); camino = buscador.buscar(); t1 = time()
        if camino:
            coste_cu.append(coste_camino(camino))
            prof_cu.append(buscador.profundidad_solucion)
            gen_cu.append(buscador.n_generados)
            exp_cu.append(buscador.n_expandidos)
            expl_cu.append(buscador.n_explorados)
            t_cu.append(t1 - t0)

        # Primero el Mejor
        buscador = PrimeroMejor(problema)
        t0 = time(); camino = buscador.buscar(); t1 = time()
        if camino:
            coste_mp.append(coste_camino(camino))
            prof_mp.append(buscador.profundidad_solucion)
            gen_mp.append(buscador.n_generados)
            exp_mp.append(buscador.n_expandidos)
            expl_mp.append(buscador.n_explorados)
            t_mp.append(t1 - t0)

        # A*
        buscador = AEstrella(problema)
        t0 = time(); camino = buscador.buscar(); t1 = time()
        if camino:
            coste_a.append(coste_camino(camino))
            prof_a.append(buscador.profundidad_solucion)
            gen_a.append(buscador.n_generados)
            exp_a.append(buscador.n_expandidos)
            expl_a.append(buscador.n_explorados)
            t_a.append(t1 - t0)

    print("\n=========== ESCENARIO 1: Mapas moderados (40 ciudades) ===========")
    imprimir_medias_y_desviaciones("Coste Uniforme", coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu, N_MAPAS)
    imprimir_medias_y_desviaciones("Primero el Mejor", coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp, N_MAPAS)
    imprimir_medias_y_desviaciones("A*", coste_a, prof_a, gen_a, exp_a, expl_a, t_a, N_MAPAS)

def escenario_2_mapas_grandes():
    """
    Escenario 2:
    - 10 mapas grandes
    - 120 ciudades
    - conectividad media
    - Origen = C1, Destino = C120
    """

    N_MAPAS = 10
    seeds = [2000 + i for i in range(N_MAPAS)]

    coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu = ([] for _ in range(6))
    coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp = ([] for _ in range(6))
    coste_a,  prof_a,  gen_a,  exp_a,  expl_a,  t_a  = ([] for _ in range(6))

    for seed in seeds:
        random.seed(seed)

        cities, roads = generate_map(
            seed=seed,
            n_cities=120,
            max_neighbours=6,   # misma conectividad que escenario 1
            radious=0.15
        )

        problema = Problema(cities, roads, "C1", "C120")

        # Coste Uniforme
        buscador = CosteUniforme(problema)
        t0 = time(); camino = buscador.buscar(); t1 = time()
        if camino:
            coste_cu.append(coste_camino(camino))
            prof_cu.append(buscador.profundidad_solucion)
            gen_cu.append(buscador.n_generados)
            exp_cu.append(buscador.n_expandidos)
            expl_cu.append(buscador.n_explorados)
            t_cu.append(t1 - t0)

        # Primero el Mejor
        buscador = PrimeroMejor(problema)
        t0 = time(); camino = buscador.buscar(); t1 = time()
        if camino:
            coste_mp.append(coste_camino(camino))
            prof_mp.append(buscador.profundidad_solucion)
            gen_mp.append(buscador.n_generados)
            exp_mp.append(buscador.n_expandidos)
            expl_mp.append(buscador.n_explorados)
            t_mp.append(t1 - t0)

        # A*
        buscador = AEstrella(problema)
        t0 = time(); camino = buscador.buscar(); t1 = time()
        if camino:
            coste_a.append(coste_camino(camino))
            prof_a.append(buscador.profundidad_solucion)
            gen_a.append(buscador.n_generados)
            exp_a.append(buscador.n_expandidos)
            expl_a.append(buscador.n_explorados)
            t_a.append(t1 - t0)

    print("\n=========== ESCENARIO 2: Mapas grandes (120 ciudades) ===========")
    imprimir_medias_y_desviaciones("Coste Uniforme", coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu, N_MAPAS)
    imprimir_medias_y_desviaciones("Primero el Mejor", coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp, N_MAPAS)
    imprimir_medias_y_desviaciones("A*", coste_a, prof_a, gen_a, exp_a, expl_a, t_a, N_MAPAS)



def coste_camino(camino):
    """Suma los costes de las acciones de un camino (lista de Accion)."""
    if camino is None:
        return float("inf")
    return sum(a.coste for a in camino)


def imprimir_medias_y_desviaciones(nombre,
                                    costes, profundidades,
                                    generados, expandidos, explorados,
                                    tiempos, n_pruebas):
    print(f"\n=== {nombre} ===")

    if len(costes) == 0:
        print("No se ha encontrado solución en ninguna de las pruebas.")
        print(f"Éxitos: 0/{n_pruebas}")
        return

    print(f"Número de éxitos: {len(costes)} / {n_pruebas}")

    print("Media del coste:              ", statistics.mean(costes))
    print("Desviación del coste:         ", statistics.stdev(costes) if len(costes) > 1 else 0.0)

    print("Media de la profundidad:      ", statistics.mean(profundidades))
    print("Desviación de la profundidad: ", statistics.stdev(profundidades) if len(profundidades) > 1 else 0.0)

    print("Media de nodos generados:     ", statistics.mean(generados))
    print("Desviación de nodos generados:", statistics.stdev(generados) if len(generados) > 1 else 0.0)

    print("Media de nodos expandidos:    ", statistics.mean(expandidos))
    print("Desviación de nodos expandidos:",
          statistics.stdev(expandidos) if len(expandidos) > 1 else 0.0)

    print("Media de nodos explorados:    ", statistics.mean(explorados))
    print("Desviación de nodos explorados:",
          statistics.stdev(explorados) if len(explorados) > 1 else 0.0)

    print("Media de tiempo (s):          ", statistics.mean(tiempos))
    print("Desviación de tiempo (s):     ",
          statistics.stdev(tiempos) if len(tiempos) > 1 else 0.0)



def escenario_A_alta_conectividad():
    """
    Escenario A:
    - 10 mapas con ALTA conectividad
    - 40 ciudades
    - max_neighbours alto
    - radious grande
    - Origen = C1, Destino = C40
    """

    import statistics
    from time import perf_counter as time
    import random

    N_MAPAS = 10
    seeds = [3000 + i for i in range(N_MAPAS)]

    # Métricas por algoritmo
    coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu = ([] for _ in range(6))
    coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp = ([] for _ in range(6))
    coste_a,  prof_a,  gen_a,  exp_a,  expl_a,  t_a  = ([] for _ in range(6))

    for seed in seeds:
        random.seed(seed)

        # Mapa pequeño pero MUY conectado
        cities, roads = generate_map(
            seed=seed,
            n_cities=40,
            max_neighbours=12,  # muchos vecinos por ciudad
            radious=0.25        # radio grande → muchas carreteras
        )

        origen_id = "C1"
        destino_id = "C40"

        problema = Problema(cities, roads, origen_id, destino_id)

        # ---- Coste Uniforme ----
        buscador = CosteUniforme(problema)
        t0 = time()
        camino = buscador.buscar()
        t1 = time()
        if camino is not None:
            coste_cu.append(coste_camino(camino))
            prof_cu.append(buscador.profundidad_solucion)
            gen_cu.append(buscador.n_generados)
            exp_cu.append(buscador.n_expandidos)
            expl_cu.append(buscador.n_explorados)
            t_cu.append(t1 - t0)

        # ---- Primero el Mejor ----
        buscador = PrimeroMejor(problema)
        t0 = time()
        camino = buscador.buscar()
        t1 = time()
        if camino is not None:
            coste_mp.append(coste_camino(camino))
            prof_mp.append(buscador.profundidad_solucion)
            gen_mp.append(buscador.n_generados)
            exp_mp.append(buscador.n_expandidos)
            expl_mp.append(buscador.n_explorados)
            t_mp.append(t1 - t0)

        # ---- A* ----
        buscador = AEstrella(problema)
        t0 = time()
        camino = buscador.buscar()
        t1 = time()
        if camino is not None:
            coste_a.append(coste_camino(camino))
            prof_a.append(buscador.profundidad_solucion)
            gen_a.append(buscador.n_generados)
            exp_a.append(buscador.n_expandidos)
            expl_a.append(buscador.n_explorados)
            t_a.append(t1 - t0)

    print("\n=========== ESCENARIO A: Alta conectividad (40 ciudades) ===========")

    imprimir_medias_y_desviaciones(
        "Coste Uniforme", coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu, N_MAPAS
    )
    imprimir_medias_y_desviaciones(
        "Primero el Mejor", coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp, N_MAPAS
    )
    imprimir_medias_y_desviaciones(
        "A*", coste_a, prof_a, gen_a, exp_a, expl_a, t_a, N_MAPAS
    )
def escenario_3_desafiante():
    """
    Escenario 3 (desafiante):
    - 10 mapas con BAJA conectividad
    - 60 ciudades
    - max_neighbours = 3
    - radious = 0.06 (pocas carreteras)
    - Origen = C1, Destino = C60
    """

    import statistics
    from time import perf_counter as time
    import random

    N_MAPAS = 10
    seeds = [4000 + i for i in range(N_MAPAS)]

    # Métricas por algoritmo
    coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu = ([] for _ in range(6))
    coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp = ([] for _ in range(6))
    coste_a,  prof_a,  gen_a,  exp_a,  expl_a,  t_a  = ([] for _ in range(6))

    for seed in seeds:
        random.seed(seed)

        cities, roads = generate_map(
            seed=seed,
            n_cities=60,
            max_neighbours=4,
            radious=0.1
        )

        origen_id = "C1"
        destino_id = "C60"

        problema = Problema(cities, roads, origen_id, destino_id)

        # ---- Coste Uniforme ----
        buscador = CosteUniforme(problema)
        t0 = time()
        camino = buscador.buscar()
        t1 = time()
        if camino is not None:
            coste_cu.append(coste_camino(camino))
            prof_cu.append(buscador.profundidad_solucion)
            gen_cu.append(buscador.n_generados)
            exp_cu.append(buscador.n_expandidos)
            expl_cu.append(buscador.n_explorados)
            t_cu.append(t1 - t0)

        # ---- Primero el Mejor ----
        buscador = PrimeroMejor(problema)
        t0 = time()
        camino = buscador.buscar()
        t1 = time()
        if camino is not None:
            coste_mp.append(coste_camino(camino))
            prof_mp.append(buscador.profundidad_solucion)
            gen_mp.append(buscador.n_generados)
            exp_mp.append(buscador.n_expandidos)
            expl_mp.append(buscador.n_explorados)
            t_mp.append(t1 - t0)

        # ---- A* ----
        buscador = AEstrella(problema)
        t0 = time()
        camino = buscador.buscar()
        
        
        t1 = time()
        if camino is not None:
            coste_a.append(coste_camino(camino))
            prof_a.append(buscador.profundidad_solucion)
            gen_a.append(buscador.n_generados)
            exp_a.append(buscador.n_expandidos)
            expl_a.append(buscador.n_explorados)
            t_a.append(t1 - t0)

    print("\n=========== ESCENARIO 3: Desafiante (baja conectividad, 60 ciudades) ===========")

    imprimir_medias_y_desviaciones(
        "Coste Uniforme", coste_cu, prof_cu, gen_cu, exp_cu, expl_cu, t_cu, N_MAPAS
    )
    imprimir_medias_y_desviaciones(
        "Primero el Mejor", coste_mp, prof_mp, gen_mp, exp_mp, expl_mp, t_mp, N_MAPAS
    )
    imprimir_medias_y_desviaciones(
        "A*", coste_a, prof_a, gen_a, exp_a, expl_a, t_a, N_MAPAS
    )
    

```
